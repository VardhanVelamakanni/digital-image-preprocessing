```python
from PIL import Image, ImageDraw
def generate_vertical_stripes():
    img = Image.new("RGB", (1024, 1024), "white")
    draw = ImageDraw.Draw(img)
    stripe_width = 20
    for x in range(0, 1024, stripe_width * 2):
        draw.rectangle([x, 0, x + stripe_width - 1, 1024], fill="black")
    img.save("vertical_stripes.png")
    print("Vertical stripes image saved as 'vertical_stripes.png'.")
if __name__ == "__main__":
    generate_vertical_stripes()
```
