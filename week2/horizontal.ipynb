```python
from PIL import Image, ImageDraw

def generate_horizontal_stripes():
    
    img = Image.new("RGB", (1024, 1024), "white")
    draw = ImageDraw.Draw(img)

    
    stripe_height = 20  
    for y in range(0, 1024, stripe_height * 2):
        draw.rectangle([0, y, 1024, y + stripe_height - 1], fill="black")

    img.save("horizontal_stripes.png")
    print("Horizontal stripes image saved as 'horizontal_stripes.png'.")

if __name__ == "__main__":
    generate_horizontal_stripes()
```
