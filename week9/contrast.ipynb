```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Part 1: Create synthetic images
def create_low_contrast_image(filename="low_contrast_image_A.png", shape=(256, 256)):
    img = np.linspace(100, 150, num=shape[0]*shape[1]).reshape(shape).astype(np.uint8)
    cv2.imwrite(filename, img)

def create_high_contrast_image(filename="high_contrast_image_B.png", shape=(256, 256)):
    img = np.zeros(shape, dtype=np.uint8)
    cv2.rectangle(img, (50, 50), (200, 200), 255, -1)
    cv2.circle(img, (128, 128), 40, 180, -1)
    cv2.line(img, (0, 0), (255, 255), 120, 2)
    cv2.imwrite(filename, img)

# Create the synthetic images
create_low_contrast_image()
create_high_contrast_image()

# Part 2: Histogram Matching Function
def histogram_matching(source, template):
    old_shape = source.shape
    source = source.ravel()
    template = template.ravel()

    # Calculate histograms and CDFs
    s_values, bin_idx, s_counts = np.unique(source, return_inverse=True, return_counts=True)
    t_values, t_counts = np.unique(template, return_counts=True)

    s_quantiles = np.cumsum(s_counts).astype(np.float64)
    s_quantiles /= s_quantiles[-1]  # Normalize

    t_quantiles = np.cumsum(t_counts).astype(np.float64)
    t_quantiles /= t_quantiles[-1]  # Normalize

    # Interpolate new values
    interp_t_values = np.interp(s_quantiles, t_quantiles, t_values)
    matched = interp_t_values[bin_idx].reshape(old_shape).astype(np.uint8)

    return matched

# Part 3: Load Images and Apply Histogram Matching
source_img = cv2.imread('low_contrast_image_A.png', cv2.IMREAD_GRAYSCALE)
template_img = cv2.imread('high_contrast_image_B.png', cv2.IMREAD_GRAYSCALE)
matched_img = histogram_matching(source_img, template_img)

# Part 4: Display Images
cv2.imshow("Source (Low Contrast)", source_img)
cv2.imshow("Template (High Contrast)", template_img)
cv2.imshow("Matched Image (Enhanced Contrast)", matched_img)

print("Press any key in the image window to close...")
cv2.waitKey(0)
cv2.destroyAllWindows()

# Part 5: Save Output
cv2.imwrite("contrast_enhanced_image.png", matched_img)
```
