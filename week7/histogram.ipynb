```python
import cv2
import numpy as np

def histogram_equalisation(input_image):
    gray = cv2.cvtColor(input_image, cv2.COLOR_BGR2GRAY)
    equalized = cv2.equalizeHist(gray)
    return equalized

# Load image
image = cv2.imread(r"D:\AP22110010458\digital-image-preprocessing\assets\photo.jpg")

# Apply histogram equalization
enhanced_image = histogram_equalisation(image)

# Save the result
cv2.imwrite('enhanced_image.jpg', enhanced_image)
```
