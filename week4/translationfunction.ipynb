```python
#Create a 1000 x 1000 image as in the attached image. It is noted that the white box size in the image is 100x100 pixels.
#create a function "translate_white_box(image,tx,ty)' to translate the white box 10 pixels right and 20 pixels down.
import numpy as np
import cv2
import matplotlib.pyplot as plt

def create_base_image(image_size=1000, box_size=100):
    image = np.zeros((image_size, image_size), dtype=np.uint8)
    center = image_size // 2
    half_box = box_size // 2
    image[center - half_box : center + half_box, center - half_box : center + half_box] = 255
    return image
def translate_white_box(image, tx, ty):
    rows, cols = image.shape
    translation_matrix = np.float32([[1, 0, tx], [0, 1, ty]])
    translated_image = cv2.warpAffine(image, translation_matrix, (cols, rows))
    return translated_image

image = create_base_image()
translated_image = translate_white_box(image, 10, 20)
plt.imshow(translated_image, cmap="gray")
plt.title("Translated Image (Right 10, Down 20)")
plt.axis("off")
plt.show()
```
